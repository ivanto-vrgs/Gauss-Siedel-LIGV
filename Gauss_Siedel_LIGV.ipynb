```python
import numpy as np
#definimos lo que necesite el método, A (matriz cuadrada), b (vector columna)
#x0 (vector cero), tolerancia (tol) y N (iteraciones).
def gauss_seidel(A, b, x0=None, tol=1e-6, N=100):
#1.
    k=1
    n = len(b)
    x = np.zeros_like(b, dtype=float) if x0 is None else x0.astype(float)
#2.

    for k in range(1, N + 1):
      #3.
        x_old = x.copy()
        for i in range(n):

            sigma1 = np.dot(A[i, :i], x[:i])
            sigma2 = np.dot(A[i, i+1:], x_old[i+1:])
            x[i] = (b[i] - sigma1 - sigma2) / A[i, i]
        #4.
        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            print(f"El procedimiento fue exitoso, convirgiendo en {k} iteraciones.")
            return x
   #Si se pasan las 100 iteraciones:
    print("Número de iteraciones excedido, el procedimiento no fue exitoso.")
    return x

#  Ejercicio del problemario: 3(c)
A = np.array([[2, -1, 0],
              [-1, 2, -1],
               [0, -1, 2]])
#asignamos un vector columna
b = np.array([4, 8, 12])
#la primera iteración es el vector cero
x0 = np.zeros(3)

solucion = gauss_seidel(A, b, x0, tol=1e-6, N=100)
print("Solución aproximada:", solucion)
```
